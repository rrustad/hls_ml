```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import ml

model_exists = False

# Initialize the WorkspaceClient
client = WorkspaceClient()

# Check if the model exists using the Databricks SDK
models = client.registered_models.list()
if any(model.full_name == model_name for model in models):
    model_exists = True
model_exists
```
