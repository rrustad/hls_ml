```python
from databricks.sdk import WorkspaceClient
import pyspark.sql.functions as f
from datetime import datetime, timedelta
```

```python
w = WorkspaceClient()
```

```python
while True:
  admissions = spark.table(f'kp_catalog.mimic_incr.admissions')
  max_adm_date = admissions.select(f.max(f.col('admittime'))).collect()[0][0]

  print(max_adm_date)
  if max_adm_date < datetime.today().replace(hour=0, minute=0, second=0, microsecond=0):
    j = w.jobs.run_now_and_wait(job_id=984999643617256,timeout=timedelta(hours=1,minutes=40))
  else:
    break
```
